```python
silver_table_path = "silver.silver_bird_conservation_areas_cleaned"
```

```python
df_silver = spark.read.table(f"{silver_table_path}")

display(df_silver.limit(5))
```

```python
### Get the Number of Records per Number of Birds Conservation Areas

df_counts = spark.sql("""
    SELECT Name, COUNT(*) AS Record_Count
    FROM silver.silver_bird_conservation_areas_cleaned
    GROUP BY Name
    ORDER BY Record_Count DESC""")

display(df_counts)
```

```python
### Create Gold Schema if not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
(df_counts.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("gold.gold_bird_conservation_areas_record_counts"))
```

```python
%sql
SELECT * FROM gold.gold_bird_conservation_areas_record_counts
```
