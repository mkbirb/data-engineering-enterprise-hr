**Notes:**
- There will be a Filter Value, where when set would retrieve all of the records that would include this Filter Value



```python
df_silver = spark.read.table("silver.silver_landbird_habitat_transformed")


display(df_silver.limit(5))
```

```python
### Retrieve the Filter Values from Parameters
dbutils.widgets.text("filter_value", "")
filter_value = dbutils.widgets.get("filter_value")
```

```python
from pyspark.sql.functions import col

df_gold = df_silver.filter(
    (col("Tree_Size_Class_Min") <= filter_value) 
    & ((col("Tree_Size_Class_Max") >= filter_value) | (col("Tree_Size_Class_Max").isNull())
))

display(df_gold)
```

```python
### Create Gold Schema if not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
### Save the Table to Catalog
(df_gold.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"gold.gold_landbird_habitat_park_tree_{filter_value}_cm_site_counts")
)
```

```python
display(spark.sql(f"""
    SELECT *
    FROM gold.gold_landbird_habitat_park_tree_{filter_value}_cm_site_counts
"""))
```
