```python
df_silver = spark.read.table("silver.silver_landbird_habitat_transformed")


display(df_silver.limit(5))
```

```python
from pyspark.sql.functions import countDistinct

df_gold_counts = df_silver.groupBy("Park_Code", "Site_Code").agg(countDistinct("Location_Code").alias("Location_Count")).orderBy("Location_Count", ascending=False)
```

```python
display(df_gold_counts)
```

```python
### Create Gold Schema if not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
### Save the Table to catalog
(df_gold_counts.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("gold.gold_landbird_habitat_park_sites_counts")
)
```

```python
%sql
SELECT * FROM gold.gold_landbird_habitat_park_sites_counts
```
