```python
df_silver = spark.read.table("silver.silver_landbird_habitat_transformed")


display(df_silver.limit(5))
```

```python
### Retrieve the Filter Values

dbutils.widgets.text("filter_day", "")
dbutils.widgets.text("filter_month", "")
dbutils.widgets.text("filter_year", "")

filter_day = int(dbutils.widgets.get("filter_day"))
filter_month = int(dbutils.widgets.get("filter_month"))
filter_year = int(dbutils.widgets.get("filter_year"))
```

```python
from pyspark.sql.functions import year, month, dayofmonth, col

df_gold_filtered = df_silver.filter(
    (year(col("Survey_Date")) == filter_year) &
    (month(col("Survey_Date")) == filter_month) &
    (dayofmonth(col("Survey_Date")) == filter_day
))
```

```python
display(df_gold_filtered)
```

```python
### Create Gold Schema if not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
date_str = f"{filter_year}{filter_month:02d}{filter_day:02d}"
table_name = f"gold.gold_landbird_habitat_park_sites_counts_{date_str}"

(df_gold_filtered.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(table_name)
)
```
