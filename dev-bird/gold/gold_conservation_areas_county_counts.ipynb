```python
silver_table_path = "silver.silver_bird_conservation_areas_cleaned"
```

```python
df_silver = spark.read.table(f"{silver_table_path}")

display(df_silver.limit(5))
```

```python
df_county_counts = spark.sql("""
    SELECT County, COUNT(*) as County_Count
    FROM silver.silver_bird_conservation_areas_cleaned
    GROUP BY County 
    ORDER BY County_Count DESC""")

display(df_county_counts)
```

```python
### Create Gold Schema if not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
(df_county_counts.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("gold.gold_bird_conservation_areas_county_counts"))
```
