```python
silver_table_path = "silver.silver_bird_conservation_areas_cleaned"
```

```python
df_silver = spark.read.table(f"{silver_table_path}")

display(df_silver.limit(5))
```

```python
df_area_of_interest_counts = spark.sql("""
    SELECT Areas_of_Interest, Count(*) AS Count
    FROM silver.silver_bird_conservation_areas_cleaned
    GROUP BY Areas_of_Interest
    ORDER BY Count DESC """)


display(df_area_of_interest_counts)
```

```python
### Create Gold Schema if does not exist
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
(df_area_of_interest_counts.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("gold.gold_bird_conservation_areas_area_of_interest_counts"))
```

```python
%sql
SELECT * FROM gold.gold_bird_conservation_areas_area_of_interest_counts
```
